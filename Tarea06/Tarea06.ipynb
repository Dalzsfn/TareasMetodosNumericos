##### Determine el orden de la mejor aproximación para las siguientes funciones, usando la Serie de Taylor y el Polinomio de Lagrange:

1. $$ f(x) =\frac{1}{25x^2+1}, x_0 = 0$$

```python
import numpy as np 
import matplotlib.pyplot as plt
```

* Serie de Taylor:

$f'(x) = -\frac{50x}{(25x^2+1)^{2}}$ <br> <br>

$f''(x) = -\frac{50}{(25x^2+1)^{2}} + \frac{5000x^2}{(25x^2+1)^{3}}$ <br> <br>

$f'''(x) = \frac{15000x}{(25x^2+1)^{3}} - \frac{750000x^3}{(25x^2+1)^{4}}$ <br> <br>

$f^{(4)}(x) = \frac{15000}{(25x^2+1)^{3}} - \frac{4500000x^2}{(25x^2+1)^{4}} + \frac{150000000x^4}{(25x^2+1)^{5}} $

Quedándonos como serie de Taylor de orden 4 con $x_0 = 0$ <br> $$f(0) \approx 1-25x^2+625x^4 $$ así, graficamos cada aproximación

$T_0(x) = 1$

```python
def f(x):
    return 1/(25*x**2 + 1)

def T0(x):
    return np.ones_like(x)

x = np.linspace(-1, 1, 400)

plt.figure()
plt.plot(x, f(x), label="f(x)")
plt.plot(x, T0(x), label="T0(x) = 1", linestyle="--")
plt.title("Aproximación de Taylor – Orden 0")
plt.xlabel("x")
plt.ylabel("Valor")
plt.legend()
plt.grid(True)
plt.show()
```

$T_2= 1-25x^2$

```python
def f(x):
    return 1/(25*x**2 + 1)

def T2(x):
    return 1 - 25*x**2

x = np.linspace(-1, 1, 400)

plt.figure()
plt.plot(x, f(x), label="f(x)")
plt.plot(x, T2(x), label="T2(x) = 1 - 25x²", linestyle="--")
plt.title("Aproximación de Taylor – Orden 2")
plt.xlabel("x")
plt.ylabel("Valor")
plt.ylim (0,1)   
plt.legend()
plt.grid(True)
plt.show()
```

$T_4 = 1-25x^2+625x^4$

```python
def f(x):
    return 1/(25*x**2 + 1)

def T4(x):
    return 1 - 25*x**2 + 625*x**4

x = np.linspace(-1, 1, 400)

plt.figure()
plt.plot(x, f(x), label="f(x)")
plt.plot(x, T4(x), label="T4(x) = 1 - 25x² + 625x⁴", linestyle="--")
plt.title("Aproximación de Taylor – Orden 4")
plt.xlabel("x")
plt.ylabel("Valor")
plt.ylim (0,1)   
plt.legend()
plt.grid(True)
plt.show()
```

* Polinomio de LaGrange

Como en la serie de Taylor usamos un polinomio de orden 4, para tener un polinomio de orden 4 en el polinomio de LaGrange debemos usar 5 puntos: $$(0,1) ~~  (0.2 , 0.5) ~~(-0.2, 0.5) ~~(0.6, 0.1)~~ (-0.6, 0.1)$$

$$P(x) = L_0(x) + 0.5L_1(x) + 0.5L_2(x) + 0.1L_3(x) + 0.1L_4(x)$$

$L_0 = \frac{(x-0.2)(x+0.2)(x-0.6)(x+0.6)}{0.0144}$ <br> <br>
$L_1 = - \frac{x(x+0.2)(x-0.6)(x+0.6)}{0.0256}$ <br> <br>
$L_2= - \frac{x(x-0.2)(x-0.6)(x+0.6)}{0.0256}$ <br> <br>
$L_3=  \frac{x(x-0.2)(x+0.2)(x+0.6)}{0.2304}$ <br> <br>
$L_4=  \frac{x(x-0.2)(x+0.2)(x-0.6)}{0.2304}$ <br> <br>
$P(x) = \frac{(x-0.2)(x+0.2)(x-0.6)(x+0.6)}{0.0144} - \frac{1}{2} (\frac{x(x+0.2)(x-0.6)(x+0.6)}{0.0256} ) - \frac{1}{2} (\frac{x(x-0.2)(x-0.6)(x+0.6)}{0.0256}) + \frac{1}{10} (\frac{x(x-0.2)(x+0.2)(x+0.6)}{0.2304}) + \frac{1}{10} (\frac{x(x-0.2)(x+0.2)(x-0.6)}{0.2304})$ <br> <br>
$$P(x) = 31.25x^4-13.75x^2+1$$

$P_0(x) = 1$

```python
def f(x):
    return 1/(25*x**2 + 1)

def T0(x):
    return np.ones_like(x)

x = np.linspace(-1, 1, 400)

plt.figure()
plt.plot(x, f(x), label="f(x)")
plt.plot(x, T0(x), label="P0(x) = 1", linestyle="--")
plt.title("Polinomio de Lagrange grado 0")
plt.xlabel("x")
plt.ylabel("Valor")
plt.legend()
plt.grid(True)
plt.show()
```

$P_2(x) = 1 - 13.75x^2$

```python
def f(x):
    return 1/(25*x**2 + 1)

def T2(x):
    return 1 - 13.75*x**2

x = np.linspace(-1, 1, 400)

plt.figure()
plt.plot(x, f(x), label="f(x)")
plt.plot(x, T2(x), label="P0(x) = 1 - 13.75x²", linestyle="--")
plt.title("Polinomio de Lagrange grado 2")
plt.xlabel("x")
plt.ylabel("Valor")
plt.ylim (0,1)   
plt.legend()
plt.grid(True)
plt.show()
```

$P_4(x) = 1 - 13.75x^2 + 31.25x^4$

```python
def f(x):
    return 1/(25*x**2 + 1)

def P(x):
    return 31.25*x**4 - 13.75*x**2 + 1

x = np.linspace(-1, 1, 400)

plt.figure()
plt.plot(x, f(x), label="f(x)")
plt.plot(x, P(x), label="P(x) = 1 - 13.75x² + 31.25x⁴", linestyle="--")
plt.title("Polinomio de Interpolación de Lagrange")
plt.xlabel("x")
plt.ylabel("Valor")
plt.axhline(y=0, color='black')
plt.ylim (-1,1)   
plt.legend()
plt.grid(True)
plt.show()
```

* Serie de Taylor

$f'(x) = \frac{1}{1+x^2}$ <br><br>

$f^{(4)}(x) = \frac{24x}{(1+x^2)^3} - \frac{48x^3}{(1+x^2)^4}$ <br><br>
$f^{(5)}(x) = \frac{24}{(1+x^2)^3} - \frac{288x^2}{(1+x^2)^4} + \frac{384x^4}{(1+x^2)^5}$ <br><br>
Quedándonos como serie de Taylor con orden 5: $$f(1) \approx \frac{\pi}{4} + \frac{1}{2}(x-1) - \frac{1}{4}(x-1)^2 + \frac{1}{12}(x-1)^3 - \frac{1}{40}(x-1)^5$$


Así, graficamos cada aproximación

$T_0(x) = \frac{\pi}{4}$
