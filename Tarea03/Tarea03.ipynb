# 1. Utilice aritmética de corte de tres dígitos para calcular las siguientes sumas. Para cada parte, ¿qué método es más preciso y por qué?

```python
import math
import numpy as np
```

Definimos una función para truncar a n cifras significativas

```python
def truncar_cifras_significativas(x, n):
    if x == 0:
        return 0
    orden = math.floor(math.log10(abs(x)))
    factor = 10 ** (orden - n + 1)
    return math.trunc(x / factor) * factor
```

Definimos las función de sumatoria para calcular el valor real

```python
def sumatoria_real(funcion, inicio, fin):
    suma = 0
    for i in range(inicio, fin + 1):
        suma += funcion(i)
    return suma
```

```python
sum1 = sumatoria_real(lambda x: 1/x**2, 1, 10)
sum1
```

```python
sum2 = sumatoria_real(lambda x: 1/x**3, 1, 10)
sum2
```

También definimos la función de sumatoria truncada en python

```python
def sumatoria(funcion, inicio, fin, inverso=False):
    suma = 0
    if inverso:
        rango = range(fin, inicio - 1, -1)
    else:
        rango = range(inicio, fin + 1)
    
    for i in rango:
        termino = truncar_cifras_significativas(funcion(i), 3)
        suma = truncar_cifras_significativas(suma + termino, 3)
    return suma
```

También definimos las funciones de error absoluto y error relativo

```python
def e_abs(aprox, real):
    return abs(aprox-real)
```

```python
def e_rel(aprox, real):
    return abs(aprox-real)/abs(real)
```

# 2. La serie de Maclaurin para la función arcotange converge para $-1 < x \le 1$ y está dada por $$arctan x = \lim_{x\to \infty} P_n(x) = \lim_{n\to \infty} \sum_{i=1}^n(-1)^{i+1}\frac{x^{2i-1}}{2i-1}$$
### a) Utilice el hecho de que $tan \pi/4 = 1$ para determinar el número n de términos de la serie que se necesita sumar para garantizar que $|4P_n(1)-\pi| < 10^{-3}$

```python
def maclaurin_arctan_approx(n):
    return sum((-1)**(i + 1) / (2 * i - 1) for i in range(1, n + 1))

n_terms = 0
precision = 1e-3
difference = 1

while difference >= precision:
    n_terms += 1
    current_approximation = 4 * maclaurin_arctan_approx(n_terms)
    difference = abs(current_approximation - math.pi)

print(f"Número de términos necesarios para alcanzar la precisión de 10^-3: {n_terms}")
```
