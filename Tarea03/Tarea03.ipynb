```python
import math
import numpy as np
```

Definimos una función para truncar a n cifras significativas

```python
def truncar_cifras_significativas(x, n):
    if x == 0:
        return 0
    orden = math.floor(math.log10(abs(x)))
    factor = 10 ** (orden - n + 1)
    return math.trunc(x / factor) * factor
```

Definimos las función de sumatoria para calcular el valor real

```python
def sumatoria_real(funcion, inicio, fin):
    suma = 0
    for i in range(inicio, fin + 1):
        suma += funcion(i)
    return suma
```

```python
sum1 = sumatoria_real(lambda x: 1/x**2, 1, 10)
sum1
```

```python
sum2 = sumatoria_real(lambda x: 1/x**3, 1, 10)
sum2
```

También definimos la función de sumatoria truncada en python

```python
def sumatoria(funcion, inicio, fin, inverso=False):
    suma = 0
    if inverso:
        rango = range(fin, inicio - 1, -1)
    else:
        rango = range(inicio, fin + 1)
    
    for i in rango:
        termino = truncar_cifras_significativas(funcion(i), 3)
        suma = truncar_cifras_significativas(suma + termino, 3)
    return suma
```

También definimos las funciones de error absoluto y error relativo

```python
def e_abs(aprox, real):
    return abs(aprox-real)
```

```python
def e_rel(aprox, real):
    return abs(aprox-real)/abs(real)
```

# 3. Otra fórmula para calcular $\pi$ se puede deducir a partir de la identidad $\pi/4 = 4arctan\frac{1}{5} - arctan\frac{1}{239}$. Determine el número de términos que se deben sumar para garantizar una aproximación $\pi$ dentro de $10^{-3}$

```python
def arctan(x, n):
    suma = 0.0
    signo = -1
    for i in range(1, n+1):
        suma = suma+ (signo **(i+1))* (x**(2*i-1)) / (2*i-1)
    return suma

tolerancia = 1e-3

n = 1

while True:
    pi_aprox = 16 * arctan(1/5, n) - 4 * arctan(1/239, n)
    
    error = abs(pi_aprox - 3.141592653589793)
    
    if error < tolerancia:
        break
    n =n+1

print(f"Número de términos : {n}")
```

# 2. Las ecuaciones (1.2) y (1.3) en la sección 1.2 proporcionan formas alternativas para las raices $x_1$ y $x_2$ de $ax^2 + bx + c = 0$. Construya un algoritmo con entrada a,b,c y salida $x_1, x_2$ que calcule las raíces $x_1$ y $x_2$ (que pueden ser iguales pero con conjugados complejos) mediante la mejor fórmula para cada raíz.

```python
import cmath

def calcular_raices(a, b, c):
    discriminante = b**2 - 4 * a * c
    
    if discriminante > 0:
        if b > 0:
            x1 = (-b - discriminante**0.5) / (2 * a)
        else:
            x1 = (-b + discriminante**0.5) / (2 * a)
        x2 = c / (a * x1)
    elif discriminante == 0:
        x1 = x2 = -b / (2 * a)
    else:
        raiz_discriminante = cmath.sqrt(discriminante)
        x1 = (-b + raiz_discriminante) / (2 * a)
        x2 = (-b - raiz_discriminante) / (2 * a)
    
    return x1, x2

a, b, c = 1, -3, 2
raiz1, raiz2 = calcular_raices(a, b, c)
print(f"Las raíces son: x1 = {raiz1}, x2 = {raiz2}")
```
