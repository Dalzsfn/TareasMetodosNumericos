# 1. Calcule los errores absoluto y relativo en las aproximaciones de p por p*
$$E_{abs} = |p* - p|$$ $$E_{rel} = \frac{|p* - p|}{|p|}$$
Definimos las funciones para calcular error absoluto y error relativo

```python
import math 
```

```python
def e_abs(aprox, real):
    return abs(aprox-real)
```

```python
def e_rel(aprox, real):
    return abs(aprox-real)/abs(real)
```

### a) $$p = \pi ,   p* = 22/7$$


$$E_{abs} = |\frac{22}{7} - \pi | = 1.264 * 10^{-3}$$ $$E_{rel} = \frac{|\frac{22}{7} - \pi|}{|\pi|} = 4.025 * 10^{-4} \approx 0.04\%  $$

```python
print(e_abs(22/7, math.pi))
```

```python
print(e_rel(22/7, math.pi))
```

```python
print(e_abs(3.1416, math.pi))
```

```python
print(e_rel(math.pi, 3.1416))
```

# 3.  Encuentre  el  intervalo más largo en el que se debe encontrar 𝑝∗ para aproximarse a 𝑝 con error  relativo máximo de $$10^{−4}$$ para cada valor de 𝑝.
Dado que tenemos $$E_{rel} = \frac{|p* - p|}{|p|} <= 10^{-4}$$ Resolviendo la inecuación tenemos $$|p* - p| <= |p| * 10^{-4}$$ Y por definición de valor absoluto tenemos qué $$-|p| * 10^{-4} <= p* - p <= p * 10^{-4}$$ Sumando p a todos los lados de la inecuación tenemos $$ p - |p| * 10^{-4} <= p* <= p + |p| * 10^{-4}$$ 
Así, podemos definir la función en python: 

```python
tolerancia = 10**-4
```

```python
def calcular_intervalo(real):
    margen = abs(real) * tolerancia
    return (f'[{real - margen}, {real + margen}]')
    
```

### a) $$\pi$$
Teniendo

$$\pi - |\pi| 10^{-4} \le p^* \le \pi + |\pi| 10^{-4}$$
Y sabiendo que $$ |\pi| 10^{-4} \approx 0.000314159 $$

Tenemos que
$p^*$ debe estar en el intervalo $[3.141278, 3.141907]$ para aproximarse a $\pi$ con un error relativo máximo de $10^{-4}$.



```python
print(calcular_intervalo(math.pi))
```

### b) $$e$$
Teniendo

$$e - |e| 10^{-4} \le p^* \le e + |e| 10^{-4}$$
Y sabiendo que $$ |e| 10^{-4} \approx 0.00027183 $$

Tenemos que
$p^*$ debe estar en el intervalo $[2.718010000276199, 2.718553656641891]$ para aproximarse a $e$ con un error relativo máximo de $10^{-4}$.

```python
print(calcular_intervalo(math.e))
```

### c) $$\sqrt2$$
Teniendo

$$\sqrt2 - |\sqrt2| 10^{-4} \le p^* \le \sqrt2 + |\sqrt2| 10^{-4}$$
Y sabiendo que $$ |\sqrt2| 10^{-4} \approx 0.000141421 $$

Tenemos que
$p^*$ debe estar en el intervalo $[1.4140721410168577, 1.4143549837293325]$ para aproximarse a $\sqrt2$ con un error relativo máximo de $10^{-4}$.

```python
print(calcular_intervalo(math.sqrt(2)))
```

### d) $$\sqrt[3] {7}$$
Teniendo

$$\sqrt[3]{7} - |\sqrt[3]{7}| 10^{-4} \le p^* \le \sqrt[3]{7} + |\sqrt[3]{7}| 10^{-4}$$
Y sabiendo que $$ |\sqrt[3]{7}| 10^{-4} \approx 0.0001913 $$

# 6. El número $e$ se puede definir por medio $e = \sum_{i=0}^{\infty} (\frac{1}{n!}) $ donde $n! = n(n-1) ... 2 . 1$ para $n \neq 0 y 0! = 1$. Calcule los errores abosluto y relativo en la siguiente aproximación de $e$
